```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt 

read_df=pd.read_csv("electricity_cost_dataset.csv")
X= read_df.drop(columns=['structure type','electricity cost'])
y= read_df['electricity cost']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

predictions = model.predict(X_test)
predictions
# score = model.score(X_test, y_test)
# score
# plt.scatter(y_test,y_test, color='blue', label='Actual')
# plt.scatter(y_test, predictions, color='red', label='Predicted')
# plt.xlabel("Actual Values")
# plt.ylabel("Predictions")
# plt.title("Actual vs Predicted Electricity Costs")
# plt.legend()
# plt.show()

plt.scatter(y_test, predictions, color='red', label='Predicted vs Actual')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()],
         color='blue', linestyle='--', label='Perfect Prediction Line')

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted Electricity Costs")
plt.legend()
plt.show()
```
